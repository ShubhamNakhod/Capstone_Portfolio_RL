```python
# Importing modules

import json
import pandas as pd
from genai_explainer import GenAIExplainer, AgentContext
from src.data.data_loader import load_all_stock_data
```

```python
# Loading the RL results

with open("results/ppo_evaluation.json") as f:
    ppo_eval =json.load(f)
with open("results/ppo_training_metrics.json") as f:
    ppo_train = json.load(f)

# Loading DQN results
with open("results/dqn_evaluation.json") as f:
    dqn_eval =json.load(f)
with open("results/dqn_training_metrics.json") as f:
    dqn_train = json.load(f)

print("PPO mean_reward:", ppo_eval["mean_reward"])
print("DQN mean_reward:", dqn_eval["mean_reward"])
```

```python
# Loading recent market data

prices = load_all_stock_data("data/")
prices = prices.tail(120) # last 120 trading days
prices.head()
```

```python
# Running the Generative AI Explainer

from dotenv import load_dotenv
load_dotenv()  # it loads the .env file (with OPENAI_API_KEY)

from genai_explainer import GenAIExaplainer. AgentContext

# Using the GPT model
explainer = GenAIExplainer(model_backend="openai")

# Example equal-weight portfolio for explanation
weights_example = {"AAPL": 0.25, "MSFT": 0.25, "GOOG": 0.25, "AMZN": 0.25}

# Context builting from PPO evaluation metrics
agent_ctx = AgentContext(
    timestep=232,
    portfolio_weights=weights_example,
    reward=ppo_eval["mean_reward"],
    episode=1,
    metrics={
        "mean_return": ppo_eval["mean_return"],
        "std_reward": ppo_eval["std_reward"],
    }
)

# Generating explanation
explanation = explainer.explain(agent_ctx, prices)

print("=== Generative AI Explanation ===")
print(explanation.reasoning)
```

```python
# Saving the Explanations
import os, json

os.makedirs("results/explanations", exist_ok=True)
with open("results/explanations/ppo_summary.json", "w") as f:
    json.dump({
        "model": "PPO",
        "reasoning": explanation.reasoning,
        "signals": explanation.key_signals,
        "risk_flags": explanation.risk_flags
    }, f, indent=2)

print("Explanation saved to results/explanations/ppo_summary.json")
```
